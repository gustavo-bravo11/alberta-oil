```python
import geopandas as gpd
from pathlib import Path
```

```python
FOLDER = "../data/raw/geospatial/transmountain"

line_gdf = gpd.read_file(
    Path(FOLDER, "transmountain_existing_raw.geojson")
)

facility_gdf = gpd.read_file(
    Path(FOLDER, "transmountain_facilities_raw.geojson")
)
```

```python
line_gdf.columns
```

```python
line_gdf['pipeline_id'] = 'trans_mountain'
line_gdf['pipeline_name'] = 'Trans Mountain Pipeline'
line_gdf['geometry_role'] = 'corridor'
line_gdf['geometry_method'] = 'source_vector'
line_gdf['source_name'] = 'NRCan'

"Either:"
"-Authoritave = Official from a government site"
"-Derived = Used some sort of method to derive it"
"-Approximate = reconstructed or digitized"
line_gdf['accuracy_class'] = 'authoritative'

line_gdf['source_crs'] = line_gdf.crs.to_string() if line_gdf.crs else None
line_gdf.rename({"OBJECTID_1":"source_object_id"}, axis=1, inplace=True)

# We will validate the the only contain one so there should not be a break here
line_gdf['geometry_type'] = line_gdf.geom_type.unique().tolist()[0]
```

```python
line_gdf = line_gdf[
    [
        "pipeline_id",
        "pipeline_name",
        "source_name",
        "source_object_id",
        "source_crs",
        "geometry_role",
        "geometry_method",
        "accuracy_class",
        "geometry",
    ]
]
```
